```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

from langchain_core.prompts import ChatPromptTemplate
#  可以传字符串和 message
# templetePrompt = ChatPromptTemplate.from_messages([
#     ("system","你是专业的评论员"),
#     ("user","请评价一下{product},包括{aspect1}和{aspect2}")
# ])
# templetePrompt = ChatPromptTemplate.from_messages([
#     ("system","你是专业的评论员"),
#     ("user","请评价一下{product},包括{aspect1}和{aspect2}")
# ])
templetePrompt = ChatPromptTemplate.from_template(
    "你是专业的评论员,请评价一下{product},包括{aspect1}和{aspect2}"
)
prompt = templetePrompt.invoke(
    {
        "product": "iphone16",
        "aspect1":'参数',
        "aspect2":"性能"
    }
)
load_dotenv(verbose=True)
llm = init_chat_model(
    model="deepseek:deepseek-flash",
)
res =  llm.invoke(prompt)
print(res.content)

```
