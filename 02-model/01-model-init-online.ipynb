## 1.调用deepseek的大模型测试一下,三个参数 base_url,model_name,api_key

```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_deepseek import ChatDeepSeek
from langchain_openai import ChatOpenAI

# 读取.env中的环境变量，以.env的环境优先
load_dotenv(verbose=True)
DEEPSEEK_API_KEY =os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
llm_deepseek = ChatDeepSeek(
    model="deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    api_base=DEEPSEEK_BASE_URL,
)
response = llm_deepseek.invoke('请你介绍一下自己')
print(response)
```

## 2.openAI兼容写法

```python
import os
from langchain_openai import ChatOpenAI
load_dotenv(verbose=True)
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
llm_chatOpen=ChatOpenAI(
    model="deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)
res = llm_chatOpen.invoke('1+1=?')
print(res)
```

## 3. init_chat_model的使用

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
load_dotenv(verbose=True)
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
model = init_chat_model(
    model="deepseek:deepseek-flash",
    base_url=DEEPSEEK_BASE_URL
)
print(model.invoke('介绍下自己'))
```
