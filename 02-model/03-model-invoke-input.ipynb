## invoke 输入参数

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv(verbose=True)
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
model = init_chat_model(
    model="deepseek:deepseek-flash",
    # temperature=1.9,
    # max_tokens=100,
    base_url=DEEPSEEK_BASE_URL
)
```

```python
## 字符串
response = model.invoke('翻译下：你好世界')
print(response)
```

```python
## 传递字典列表
messages = [
    {"role":'system','content':'你是专业的程序员'},
    {"role":'user','content':'解释一下什么是面向对象语言'}
]
res = model.invoke(messages)
print(res)
```

```python
## 多轮对话
messages = [
    {"role":'system','content':'你是专业的数学老师'},
    {"role":'user','content':'1+1=？'},
    {"role":'assistant','content':'3'},
    {"role":'user','content':'我刚才问了什么问题？'},
]
res1 = model.invoke(messages)
print(res1)
```

```python
## 记忆功能
conversation = [
    {'role':'system','content':'你是一位友好的智能助手'},
    {'role':'user', 'content':'你好我叫小明'}
]
res = model.invoke(conversation)
print(f"响应1:{res.content}")
conversation.append({'role':'assistant','content':res.content})
conversation.append({'role':'user','content':'我叫什么名字？'})
print(model.invoke(conversation).content)
```

```python
## 消息对象类型
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

message = [
    SystemMessage(content='你是专业的数学专家'),
    HumanMessage(content='解释一下一元二次方程')
]

res = model.invoke(message)
print(res.content)
```

```python
## 消息对象类型 记忆
from langchain_core.messages import SystemMessage, HumanMessage,AIMessage

message = [
    SystemMessage(content='你是专业的数学专家'),
    HumanMessage(content='解释一下一元二次方程')
]

res = model.invoke(message)
print(f"ai1回复：{res.content}")
message.append(AIMessage(content=res.content))
message.append(HumanMessage(content="我刚问的什么问题"))
print(f"回复2:{model.invoke(message).content}")
```
